## Date values

- d/m/yyyy

```python
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True, errors='coerce').dt.date
```

## Table split

Create the sizes table

```python
unique_sizes = df['pizza_size'].unique()

df_sizes = pd.DataFrame({
    'pizza_size': unique_sizes
})

size_to_n_people = {'S':2, 'M':4, 'L':8, 'XL':10, 'XXL':12}

df_sizes['size_to_n_people'] = df_sizes['pizza_size'].map(size_to_n_people)

df_sizes.sort_values('size_to_n_people', inplace=True)
df_sizes.insert(0, 'size_id', range(1, len(df_sizes)+1))
df_sizes.reset_index(drop=True, inplace=True)
df_sizes
```

Create the categories table

```python
unique_category = df['pizza_category'].unique()

df_category = pd.DataFrame({
    'pizza_category': unique_category
})

df_category.insert(0, 'category_id', range(1, len(df_category)+1))
df_category.reset_index(drop=True, inplace=True)
df_category
```

Create the pizzas table

```python
df_pizza = df[['pizza_name', 'pizza_name_short', 'unit_price', 'pizza_ingredients',
               'pizza_size', 'pizza_category']]

df_pizza.drop_duplicates(inplace=True)
df_pizza.insert(0, 'pizza_id', range(1, len(df_pizza)+1))
```

```python
df_pizza = df_pizza.merge(df_category[['pizza_category', 'category_id']],
                          on = 'pizza_category',
                          how='left')
df_pizza.drop(columns=['pizza_category'], inplace=True)

df_pizza = df_pizza.merge(df_sizes[['pizza_size', 'size_id']],
                          on = 'pizza_size',
                          how = 'left')
df_pizza.drop(columns=['pizza_size'], inplace=True)
```

Create the lines (one line per pizza pero order) table

```python
df_line = df[['order_id', 'quantity', 'total_price', 'pizza_name_short']]
df_line.rename(columns={'total_price':'total_price_line'}, inplace=True)
df_line.insert(0, 'line_id', range(1, len(df_line)+1))
df_line.reset_index(drop=True, inplace=True)
```

```python
df_line = df_line.merge(df_pizza[['pizza_name_short', 'pizza_id']],
                        on = 'pizza_name_short',
                        how='left')
df_line.drop(columns=['pizza_name_short'], inplace=True)
```

Create the orders (each order can have more than one line) table

```python
df_orders = df[['order_id', 'order_date', 'order_time']]
df_orders.drop_duplicates(inplace=True)
df_orders.reset_index(drop=True, inplace=True)
```

```python
df_temp = pd.DataFrame(pd.merge(df_orders, df_line, on='order_id').groupby('order_id')['total_price_line'].sum())
df_temp.rename(columns={'total_price_line':'price_order'}, inplace=True)
df_temp.reset_index(inplace=True)
```

```python
df_orders = df_orders.merge(df_temp[['order_id', 'price_order']],
                          on = 'order_id',
                          how = 'left')
```

## Tables

We create another conenction variable due to the fact that we will now on work in the Pizza_sales_db database

```python
conn = podbc.connect("Driver={ODBC Driver 17 for SQL Server};"
                     "Server=ACER-SWIFT-3-SF\SQLEXPRESS01;"
                     "Database=Pizza_sales_db;"
                     "Trusted_Connection=yes;",
                     autocommit=True)
```

```python
cursor = conn.cursor()
```

```python
cursor.execute("IF OBJECT_ID('Categories') IS NOT NULL DROP TABLE Categories")
```

```python
cursor.execute("""
    CREATE TABLE Categories(
        category_id INT PRIMARY KEY NOT NULL,
        pizza_category NVARCHAR(50)
    );
    """)
```

```python
cursor.execute("IF OBJECT_ID('Sizes') IS NOT NULL DROP TABLE Sizes")
```

```python
cursor.execute("""
    CREATE TABLE Sizes(
        size_id INT PRIMARY KEY NOT NULL,
        pizza_size NVARCHAR(10),
        size_to_n_people INT
    );
    """)
```

```python
cursor.execute("IF OBJECT_ID('Pizzas') IS NOT NULL DROP TABLE Pizzas")
```

```python
cursor.execute("""
    CREATE TABLE Pizzas(
        pizza_id INT PRIMARY KEY NOT NULL,
        pizza_name NVARCHAR(50),
        pizza_name_short NVARCHAR(50),
        unit_price DECIMAL(10,2),
        pizza_ingredients VARCHAR(200),
        category_id INT,
        size_id INT
    );
    """)
```

```python
cursor.execute("IF OBJECT_ID('Lines') IS NOT NULL DROP TABLE Lines")
```

```python
cursor.execute("""
    CREATE TABLE Lines(
        line_id INT PRIMARY KEY NOT NULL,
        order_id INT,
        quantity INT,
        total_price_line DECIMAL(10,2),
        pizza_id INT
    );
    """)
```

```python
cursor.execute("IF OBJECT_ID('Orders') IS NOT NULL DROP TABLE Orders")
```

```python
cursor.execute("""
    CREATE TABLE Orders(
        order_id INT PRIMARY KEY NOT NULL,
        order_date DATE,
        order_time TIME(0),
        price_order DECIMAL(10,2)
    );
    """)
```

```python
cursor.close()
```

```python
cursor.execute("""
               ALTER TABLE Lines
               ADD CONSTRAINT FK_Lines_Orders
               FOREIGN KEY (order_id) REFERENCES Orders (order_id);
               """)
```
